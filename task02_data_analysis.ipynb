```python
# TASK 02 - DATA ANALYSIS

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv('/content/raw_customer_data.csv')

print("========== DATASET PREVIEW ==========")
print(df.head())
print()

# Convert numeric columns properly
df['Income'] = pd.to_numeric(
    df['Income'].astype(str).str.replace(',', '').str.strip(),
    errors='coerce'
)

df['Purchase_Amount'] = pd.to_numeric(
    df['Purchase_Amount'].astype(str).str.replace(',', '').str.strip(),
    errors='coerce'
)

# Basic information
print("========== DATASET INFORMATION ==========")
df.info()
print()

# Dataset shape
print("========== DATASET SHAPE ==========")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print()

# Missing values
print("========== MISSING VALUES ==========")
print(df.isnull().sum())
print()

# Statistical summary
print("========== STATISTICAL SUMMARY ==========")
print(df.describe())
print()

# Gender analysis
print("========== GENDER DISTRIBUTION ==========")
print(df['Gender'].value_counts())
print()

# City analysis
print("========== CITY DISTRIBUTION ==========")
print(df['City'].value_counts())
print()

# Category analysis
print("========== CATEGORY DISTRIBUTION ==========")
print(df['Category'].value_counts())
print()

# Income analysis
print("========== INCOME ANALYSIS ==========")
print("Average Income:", df['Income'].mean())
print("Highest Income:", df['Income'].max())
print("Lowest Income:", df['Income'].min())
print()

# Purchase analysis
print("========== PURCHASE ANALYSIS ==========")
print("Average Purchase Amount:", df['Purchase_Amount'].mean())
print("Total Purchase Amount:", df['Purchase_Amount'].sum())
print("Highest Purchase Amount:", df['Purchase_Amount'].max())
print("Lowest Purchase Amount:", df['Purchase_Amount'].min())
print()

# Highest spending customer
print("========== TOP CUSTOMER ==========")
top_customer = df.loc[df['Purchase_Amount'].idxmax()]
print(top_customer)
print()

# Category-wise purchase amount
print("========== CATEGORY-WISE PURCHASE ==========")
category_sales = df.groupby('Category')['Purchase_Amount'].sum().sort_values(ascending=False)
print(category_sales)
print()

# City-wise purchase amount
print("========== CITY-WISE PURCHASE ==========")
city_sales = df.groupby('City')['Purchase_Amount'].sum().sort_values(ascending=False)
print(city_sales)
print()

# -------- GRAPHS --------

# Gender distribution graph
plt.figure(figsize=(7,5))
df['Gender'].value_counts().plot(kind='bar')
plt.title('Gender Distribution')
plt.xlabel('Gender')
plt.ylabel('Number of Customers')
plt.tight_layout()
plt.show()

# City distribution graph
plt.figure(figsize=(7,5))
df['City'].value_counts().plot(kind='bar')
plt.title('Customer Distribution by City')
plt.xlabel('City')
plt.ylabel('Number of Customers')
plt.tight_layout()
plt.show()

# Category distribution graph
plt.figure(figsize=(7,5))
df['Category'].value_counts().plot(kind='bar')
plt.title('Category Distribution')
plt.xlabel('Category')
plt.ylabel('Number of Customers')
plt.tight_layout()
plt.show()

# Category-wise sales graph
plt.figure(figsize=(7,5))
category_sales.plot(kind='bar')
plt.title('Purchase Amount by Category')
plt.xlabel('Category')
plt.ylabel('Total Purchase Amount')
plt.tight_layout()
plt.show()

# City-wise sales graph
plt.figure(figsize=(7,5))
city_sales.plot(kind='bar')
plt.title('Purchase Amount by City')
plt.xlabel('City')
plt.ylabel('Total Purchase Amount')
plt.tight_layout()
plt.show()

print("========== TASK 02 COMPLETED ==========")
```
